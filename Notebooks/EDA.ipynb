```python
# ==============================================================================
# CELDA 1: Importación de librerías
# ==============================================================================
import pandas as pd
import plotly.express as px

# ==============================================================================
# CELDA 2: Carga del dataset y primera inspección
# ==============================================================================
# Cargar los datos desde el archivo CSV
df = pd.read_csv('data/vehicles_us.csv')

print("--- Primeras 5 filas del conjunto de datos ---")
display(df.head())

print("\n--- Resumen del dataset y tipos de datos ---")
df.info()

print("\n--- Conteo de valores nulos por columna ---")
print(df.isnull().sum())

# ==============================================================================
# CELDA 3: Limpieza y preparación básica de datos
# ==============================================================================
# Imputamos la columna is_4wd: NaN indica que el vehículo no es 4WD (0.0)
df['is_4wd'] = df['is_4wd'].fillna(0.0)

# Convertimos la fecha de publicación a formato datetime para facilitar análisis temporales
df['date_posted'] = pd.to_datetime(df['date_posted'])

# Limpieza ligera de etiquetas
df['model'] = df['model'].str.title()

# ==============================================================================
# CELDA 4: Histograma - Distribución de Precios
# ==============================================================================
fig_price = px.histogram(
    df, 
    x='price', 
    nbins=40, 
    title='Distribución General de Precios',
    labels={'price': 'Precio ($USD)', 'count': 'Cantidad de Vehículos'},
    color_discrete_sequence=['#3366CC']
)
fig_price.update_layout(template='plotly_white', bargap=0.1)
fig_price.show()

# ==============================================================================
# CELDA 5: Dispersión - Kilometraje vs Precio según Condición
# ==============================================================================
fig_scatter = px.scatter(
    df, 
    x='odometer', 
    y='price', 
    color='condition',
    hover_data=['model', 'model_year', 'type'],
    title='Relación entre Kilometraje (Odometer) y Precio',
    labels={
        'odometer': 'Kilometraje (Millas)', 
        'price': 'Precio ($USD)', 
        'condition': 'Condición'
    },
    opacity=0.7
)
fig_scatter.update_layout(template='plotly_white')
fig_scatter.show()

# ==============================================================================
# CELDA 6: Diagrama de Cajas - Comparativa de Precio por Tipo de Vehículo
# ==============================================================================
fig_box = px.box(
    df, 
    x='type', 
    y='price', 
    color='type',
    title='Distribución de Precios por Tipo de Vehículo',
    labels={'type': 'Tipo de Vehículo', 'price': 'Precio ($USD)'}
)
fig_box.update_layout(template='plotly_white', showlegend=False)
fig_box.show()

# ==============================================================================
# CELDA 7: Histograma - Días Publicado por Tipo de Combustible
# ==============================================================================
fig_days = px.histogram(
    df, 
    x='days_listed', 
    color='fuel', 
    barmode='overlay',
    title='Tiempo de Publicación (Días) según el Tipo de Combustible',
    labels={'days_listed': 'Días Publicado', 'count': 'Cantidad de Anuncios'}
)
fig_days.update_layout(template='plotly_white')
fig_days.show()
```
